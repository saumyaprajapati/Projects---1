<h3>🌞Importing Required Dependencies</h3>

```python
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
```

```python
htmlFile=open("Flipkart_SmartPhone.html",encoding="UTF-8").read()
soup = BeautifulSoup(htmlFile,"html.parser")
```

```python
'''
🌞 For Name :-
            <div class="KzDlHZ">Apple iPhone 16 (Black, 128 GB)</div>

🌞 For Star :-
            <div class="XQDdHH">4.6</div>

🌞 For Rating :-
            <span class="Wphh3N"><span><span>16,639 Ratings&nbsp;</span><span class="hG7V+4">&amp;</span><span>&nbsp;704 Reviews</span></span></span>

🌞 For ROM :-
            <ul class="G4BRas">
            <li class="J+igdf">128 GB ROM</li>
            </ul>

🌞For Display_Inch :-
            <ul class="G4BRas"><li class="J+igdf">15.49 cm (6.1 inch) Super Retina XDR Display</li></ul>

🌞For New Price :-
            <div class="hl05eU"><div class="Nx9bqj _4b5DiR">₹74,900</div><div class="yRaY8j ZYYwLA">₹79,900</div><div class="UkUFwK"><span>6% off</span></div></div>

🌞For Old Price :-
            <div class="hl05eU"><div class="yRaY8j ZYYwLA">₹79,900</div></div>

🌞For Discount :-
            <div class="UkUFwK">
            <span>6% off</span>
            </div>
'''
```

```python
# 🤔 Extracting Separate Html Block Of Each Smartphone
SmartPhoneBlock=soup.find_all("div",{"class":"tUxRFH"})
```

```python
print(SmartPhoneBlock[0].prettify())
```

```python
# 🤔 Extracting Smartphone Name
SmartPhoneName=SmartPhoneBlock[0].find("div",{"class":"KzDlHZ"}).text.split("(")[0].strip()
print(SmartPhoneName)
```

```python
# 🤔 Extracting Star 
Star=float(SmartPhoneBlock[0].find("div",{"class":"XQDdHH"}).text)
print(Star)
```

```python
# 🤔 Extracting Rating
Rating=SmartPhoneBlock[0].find("span",{"class":"Wphh3N"}).text.split(" ")[0].strip().split(",")
Rating=int(Rating[0])*1000 + int(Rating[1])
print(Rating)
```

```python
# 🤔 Extracting Smart Phone Detail Block
SmartPhoneDetail=SmartPhoneBlock[0].find("li",{"class" : "J+igdf"}).text
print(SmartPhoneDetail)
```

```python

# 🤔 Extracting ROM
ROM=int(SmartPhoneDetail.split("GB")[0])
print(ROM)
```

```python
# 🤔 Extracting Display_Inch
Display_Inch=float(SmartPhoneDetail.split("inch")[0].strip().split("(")[1])
print(Display_Inch)
```

```python
# 🤔 Extracting New Price
NewPrice=SmartPhoneBlock[0].find("div",{"class":"Nx9bqj _4b5DiR"}).text.strip()[1::].split(",")
NewPrice=int(NewPrice[0])*1000+int(NewPrice[1])
print(NewPrice)
```

```python
# 🤔 Extracting Old Price
OldPrice=SmartPhoneBlock[0].find("div",{"class":"yRaY8j ZYYwLA"}).text.strip()[1::].split(",")
OldPrice=float(OldPrice[0])*1000+float(OldPrice[1])
print(OldPrice)
```

```python
# 🤔 Extracting Discount
Discount=SmartPhoneBlock[0].find("div",{"class":"UkUFwK"}).text
if("%" in Discount):
    Discount=float(Discount.split("%")[0])
else: 
    Discount=np.nan
print(Discount)
```

```python
DataFrameDictionary={"Name":[],"Star":[],"Rating":[],"ROM":[],"Display_inch":[],"New_price":[],"old_price":[],"Discount":[]}
```

```python
for i in range(len(SmartPhoneBlock)):
    # 🤔 Extracting Smartphone Name
    SmartPhoneName=SmartPhoneBlock[i].find("div",{"class":"KzDlHZ"}).text.strip()
    # 🤔 Extracting Star 
    Star=float(SmartPhoneBlock[i].find("div",{"class":"XQDdHH"}).text)
    # 🤔 Extracting Rating
    Rating=SmartPhoneBlock[i].find("span",{"class":"Wphh3N"}).text.split(" ")[0].strip()
    if("," in Rating):
        Rating=Rating.split(",")
        Rating=int(Rating[0])*1000 + int(Rating[1])
    else:
        Rating=int(Rating)
    # 🤔 Extracting Smart Phone Detail Block
    SmartPhoneDetail=SmartPhoneBlock[i].find("li",{"class" : "J+igdf"}).text
    # 🤔 Extracting ROM
    ROM=int(SmartPhoneDetail.split("GB")[0])
    # 🤔 Extracting Display_Inch
    Display_Inch=float(SmartPhoneDetail.split("inch")[0].strip().split("(")[1])
    # 🤔 Extracting New Price
    NewPrice=SmartPhoneBlock[i].find("div",{"class":"Nx9bqj _4b5DiR"}).text.strip()[1::].split(",")
    NewPrice=int(NewPrice[0])*1000+int(NewPrice[1])
    # 🤔 Extracting Old Price
    OldPrice=SmartPhoneBlock[i].find("div",{"class":"yRaY8j ZYYwLA"}).text.strip()[1::].split(",")
    OldPrice=float(OldPrice[0])*1000+float(OldPrice[1])
    # 🤔 Extracting Discount
    Discount=SmartPhoneBlock[i].find("div",{"class":"UkUFwK"}).text
    if("%" in Discount):
        Discount=float(Discount.split("%")[0])
    else: 
        Discount=np.nan
    DataFrameDictionary["Name"].append(SmartPhoneName)
    DataFrameDictionary["Star"].append(Star)
    DataFrameDictionary["Rating"].append(Rating)
    DataFrameDictionary["ROM"].append(ROM)
    DataFrameDictionary["Display_inch"].append(Display_Inch)
    DataFrameDictionary["New_price"].append(NewPrice)
    DataFrameDictionary["old_price"].append(OldPrice)
    DataFrameDictionary["Discount"].append(Discount)
    print(i+1)
```

```python
df=pd.DataFrame(DataFrameDictionary)
```

```python
df.to_csv("SmartPhone.csv")
```
